```python
import numpy as np
import matplotlib.pyplot as plt


N = 4000
change_point = 2000


M = 3 #taps
mu_values = [0.0005, 0.005, 0.05]

# h = np.array([0.5, 0.25, -0.3])
h1 = np.array([0.5,0.25,-0.3])
h2 = np.array([0.8,-0.1,-0.5])
samples = np.arange(N)
seed = 11
noise_std = 1.0
rng = np.random.default_rng(seed)
k = np.arange(N)
x = rng.normal(0, noise_std, N)

use_sinus = False


#unknown system

# d = np.convolve(x,h,mode="full")[:N]

d1 = np.convolve(x, h1, mode="full")[:N]
d2 = np.convolve(x, h2, mode="full")[:N]

d = np.where(np.arange(N) < change_point, d1, d2)

#container for multiple runs of mu
results = {}

#LMS
for mu in mu_values:
    w = np.zeros(M) #inital weights are zero
    y = np.zeros(N) #adaptive filter output Y
    e = np.zeros(N) #error signal
    w_history = np.zeros((N,M))

    for k in range(M-1, N):
        #
        x_vec = x[k - M + 1:k + 1][::-1] #creates the X[k] vector 
        #extracts M samples from x, including the current samples
        #the reverses the order [::-1]
        
        #adaptive filter output
        y[k] = np.dot(w,x_vec) # W[k]*X[k] to apply the filter

        #calculate error
        e[k] = d[k] - y[k]

        #update the weights
        w = w + 2 * mu * e[k] * x_vec

        #store weights 
        w_history[k] = w

    results[mu] = {
        "y": y.copy(),
        "e": e.copy(),
        "w_history": w_history.copy()
    }


# fig, axes = plt.subplots(
#     len(mu_values), 1,
#     figsize=(12, 9),
#     sharex=True
# )

# for i, mu in enumerate(mu_values):

#     ax = axes[i]

#     ax.plot(samples, d, "k--", linewidth=0.9,
#             alpha=0.65, label="Desired d[k]")

#     ax.plot(samples, results[mu]["y"],
#             linewidth=0.9, alpha=0.8,
#             label=f"Output y[k], μ={mu}")

#     ax.axvline(change_point, color="red",
#                linestyle="--", linewidth=1.2,
#                label="System change")

#     ax.set_ylabel("Amplitude")
#     ax.set_title(f"LMS output, μ={mu}, M={M}")
#     ax.legend(loc="upper right")
#     ax.grid(True, alpha=0.3)

# axes[-1].set_xlabel("Sample k")

# fig.tight_layout()

# # ==========================================================
# # Plot 2: Weight convergence and system change
# # ==========================================================

# fig, ax = plt.subplots(figsize=(12, 6))

# linestyles = ["-", "--", ":"]
# colors = plt.cm.tab10.colors

# # Adaptive weights for each step size
# for i, mu in enumerate(mu_values):

#     for m in range(M):

#         ax.plot(
#             samples,
#             results[mu]["w_history"][:, m],
#             color=colors[m],
#             linestyle=linestyles[i % len(linestyles)],
#             linewidth=1.3,
#             label=f"w[{m}], μ={mu}"
#         )

# # True coefficients before and after system change
# for m in range(M):

#     color = colors[m]

#     ax.hlines(
#         h1[m],
#         xmin=0,
#         xmax=change_point,
#         color=color,
#         linestyle="-.",
#         linewidth=2,
#         alpha=0.65,
#         label=f"True h[{m}]"
#     )

#     ax.hlines(
#         h2[m],
#         xmin=change_point,
#         xmax=N - 1,
#         color=color,
#         linestyle="-.",
#         linewidth=2,
#         alpha=0.65
#     )

#     # Vertical transition between true coefficients
#     ax.plot(
#         [change_point, change_point],
#         [h1[m], h2[m]],
#         color=color,
#         linestyle="-.",
#         linewidth=1.5,
#         alpha=0.65
#     )

# ax.axvline(
#     change_point,
#     color="black",
#     linestyle="--",
#     linewidth=1.2,
#     alpha=0.7,
#     label="System change"
# )

# ax.set_xlabel("Sample k")
# ax.set_ylabel("Weight value")
# ax.set_title(
#     f"LMS weight convergence and system change, M={M}"
# )

# ax.set_xlim(0, N - 1)
# ax.legend(loc="best", ncol=2, fontsize=8)
# ax.grid(True, alpha=0.3)

# fig.tight_layout()
# plt.show()

```
