```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import decimate, resample_poly

fs = 1e3
#load sample
x = np.loadtxt("SampleRateData.dat")
#sample index
n = np.arange(len(x))
print(f"number of samples: {len(x)}")
# print(x[:100])

plt.figure(figsize=(12,6))
plt.xlabel("sample[n]")
plt.ylabel("magnitude")
plt.title("up/down sampling")
plt.stem(n,x)
plt.grid(True)
plt.show()
```

```python
#downsample by 2
M = 2
#load samples
y = decimate(x, M, ftype='fir', zero_phase = True)
#create new ttime axis, essentially 500Hz

ny = np.arange(len(y))

plt.figure(figsize=(12,6))
plt.stem(ny,y)
plt.xlabel("sample[n]")
plt.ylabel("magnitude")
plt.title("up/down sampling")
plt.grid(True)
plt.show()
```

```python
#upsample by 2
L = 2
z_up = np.zeros(len(y) * L)
z = resample_poly(y,up=L, down=1)

# z = np.interp(
#     np.arange(0,len(y) - 1,0.5),
#     np.arange(len(y)),
#     y
#     )

nz = np.arange(len(z))
plt.figure(figsize=(12,6))
plt.xlabel("sample[n]")
plt.ylabel("magnitude")
plt.title("up/down sampling")
plt.stem(nz,z)
plt.grid(True)
plt.show()

# plt.figure(figsize=(10, 8))

# plt.subplot(3, 1, 1)
# plt.plot(np.arange(len(x)) / 1000, x)
# plt.title("Original x[n], fs = 1000 Hz")
# plt.grid()

# plt.subplot(3, 1, 2)
# plt.plot(np.arange(len(y)) / 500, y)
# plt.title("Downsampled y[n], fs = 500 Hz")
# plt.grid()

# plt.subplot(3, 1, 3)
# plt.plot(np.arange(len(z)) / 1000, z)
# plt.title("Interpolated z[n], fs = 1000 Hz")
# plt.grid()
```
