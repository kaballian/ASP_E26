```python
import numpy as np
import matplotlib.pyplot as plt

#create the time axis
omega = np.linspace(-np.pi, np.pi, 2000)

#PSD
PSD = 20 - 16 * np.cos(3*omega)
plt.figure(figsize=(10,5))
plt.plot(omega / np.pi, PSD, linewidth=2)
plt.xlabel(r'normalized angular frequency, rad/sample')
plt.ylabel('amplitude')
plt.title('power spectral density of random stationary process')
plt.xticks(
    [-1, -2/3, -1/3, 0, 1/3, 2/3, 1],
    [r'$-\pi$', r'$-\frac{2\pi}{3}$',
     r'$-\frac{\pi}{3}$', '0',
     r'$\frac{\pi}{3}$', r'$\frac{2\pi}{3}$',
     r'$\pi$']
)

plt.grid(True)
plt.xlim(-1, 1)
plt.ylim(0, 40)
plt.show()
```

```python
import numpy as np

# Autocorrelation coefficients, l = -3,...,3
rxx = [-8, 0, 0, 20, 0, 0, -8]

# Filter impulse response
h = [5, 1]

# Convolution
ryx = np.convolve(h, rxx)

lags = np.arange(-3, 5)

for lag, value in zip(lags, ryx):
    print(f"r_yx[{lag}] = {value}")
```
